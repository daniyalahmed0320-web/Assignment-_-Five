### Q1-STUDENT :
- PRIVATE VARIABLES + GETTER // SETTER :

```python
class Student:
    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_roll_number(self):
        return self.__roll_number

    def set_roll_number(self, roll_number):
        self.__roll_number = roll_number


student = Student("Daniyal", 674222)

print("Name:", student.get_name())
print("Roll Number:", student.get_roll_number())

student.set_name("Hammad")
student.set_roll_number(677888)

print("Updated Name:", student.get_name())
print("Updated Roll Number:", student.get_roll_number())
```

### Q2- BANK ACCOUNT :
- PRIVATE BALANCE 

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print("Deposit successful")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawal successful")
        else:
            print("Insufficient balance")

    def check_balance(self):
        return self.__balance


account = BankAccount(55000)

print("Initial Balance:", account.check_balance())

account.deposit(21100)
print("Balance:", account.check_balance())

account.withdraw(17900)
print("Balance:", account.check_balance())

account.withdraw(6900)
```

### Q3- EMPLOYEE :
- PRIVATE SALARY + VALIDATION

```python
class Employee:
    def __init__(self, salary):
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary < 0:
            raise ValueError("Invalid Salary")
        self.__salary = salary


employee = Employee(65000)

print("Salary:", employee.get_salary())

employee.set_salary(120000)
print("Updated Salary:", employee.get_salary())

try:
    employee.set_salary(-15000)
except ValueError as e:
    print(e)
```

### Q13- OPERATOR OVERLOADING :

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)


num1 = Number(196)
num2 = Number(299)
num3 = Number(678)

result = num1 + num2 + num3

print("Result:", result.value)
```

### Q16- ADSTRACT CLASS :

```python
from abc import ABC, abstractmethod

class Computer(ABC):

    @abstractmethod
    def process(self):
        pass

class Laptop(Computer):
    def process(self):
        print("Laptop is processing")

class Mini_desktop(Computer):
    def process(self):
        print("Mini_desktop is processing")


laptop = Laptop()
mini_desktop = Mini_desktop()

laptop.process()
mini_desktop.process()
```

### Q18- ABSTRACT + CONCRETE METHOD :

```python
from abc import ABC, abstractmethod

class Computer(ABC):

    def start_system(self):
        print("Computer system started")

    @abstractmethod
    def shutdown(self):
        pass

class Laptop(Computer):
    
    def restart(self):
        print("Laptop is restart")

    def working(self):
        print("Laptop is working")
    
    def shutdown(self):
        print("Laptop is shutting down")

laptop = Laptop()

laptop.start_system()
laptop.restart()
laptop.working()
laptop.shutdown()
```

### Q19- ABSTRACTION + ENCAPSULATION :

```python
from abc import ABC, abstractmethod

class ATM(ABC):
    def __init__(self, balance):
        self.__balance = balance

    @abstractmethod
    def withdraw(self, amount):
        pass

    def show_balance(self):
        print("Balance:", self.__balance)


class MyATM(ATM):
    def withdraw(self, amount):
        if amount <= self._ATM__balance:
            self._ATM__balance -= amount
            print("Please collect your cash")
        else:
            print("Insufficient balance")


atm = MyATM(10000)
atm.withdraw(3000)
```

```python
1. Abstraction: Abstraction means hiding implementation details and showing only essential features to the user.
2. Encapsulation: Encapsulation means combining data and methods into a single class and controlling access to that data.
Main Difference: Abstraction hides implementation details, while Encapsulation protects data by controlling access to it.
```

### Q20- REAL WORLD PAYMENT SYSTEM :

```python
class Payment:
    def pay(self, amount):
        print("Payment processing")

class CreditCard(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using Credit Card")

class EasyPaisa(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using Easypaisa")

class Cash(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using Cash")
        
class Jazzcash(Payment):
        def pay(self, amount):
            print(f"Paid Rs. {amount} using Jazzcash")
       
def make_payment(payment_method, amount):
    payment_method.pay(amount)

card = CreditCard()
easypaisa = EasyPaisa()
jazzcash = Jazzcash()
cash = Cash()

make_payment(card, 21100)
make_payment(easypaisa, 20200)
make_payment(jazzcash, 20150)
make_payment(cash, 20000)
```
